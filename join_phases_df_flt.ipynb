```python
df0 = ld()
df0 = df0.drop(df0[df0.stat_count == 0].index)
df0.head()
```

```python
df0 = df0[df0['group_key'].str.contains('val', na=False)]
# df0 = df0[df0['group_key'].str.contains('val', na=False)]
df0 = df0[~df0['group_key'].str.contains('Sweep', na=False)]
df0.head()
```

```python
# 'faza1', 'src', 'tr_val', 'ph_nr', 'trg_var',
df0['faza1'] = df0['source_file'].astype(str).str.split('_').str[0]
df0['src'] = 'x'
df0.loc[df0['source_file'].astype(str).str.split('_').str.len() > 2, 'src'] = df0['source_file'].astype(str).str.split('_').str[1] + '_' + df0['source_file'].astype(str).str.split('_').str[2]
df0['tr_val'] = 's' # start with tr
df0.loc[df0['faza1'] == 'ph2', 'tr_val'] = 'subm'
df0.loc[df0['faza1'] == 'ph3', 'tr_val'] = 'dk'
df0.loc[df0['faza1'] == 'dk', 'tr_val'] = 'dk'
df0.loc[df0['faza1'] == 'subs', 'tr_val'] = 'sub'
df0.loc[df0['faza1'].str.contains('S', na=False), 'tr_val'] = 'sub'
df0.loc[df0['faza1'].str.contains('subs', na=False), 'tr_val'] = 'sub'
# df0.loc[df0['src'].str.contains('m', na=False), 'tr_val'] = 'm' # nie
df0['tr_val'] = df0['tr_val'] + '_' + df0['group_key'].astype(str).str.split('=').str[1] # join with val
df0['ph_nr'] = df0['target_variable'].astype(str).str[0]
df0['trg_var'] = 'DK'
df0.loc[df0['target_variable'].str.contains('SYNT', na=False), 'trg_var'] = 'SYNT'
df0.loc[df0['target_variable'].str.contains('GDA', na=False), 'trg_var'] = 'GDA'
df0.head()
```
